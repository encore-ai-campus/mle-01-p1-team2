```python
import os
import json
import pandas as pd

EVENT = r"C:\Users\Playdata\Downloads\59.반려견 성장 및 질병 관련 말뭉치 데이터\3.개방데이터\1.데이터\Training\02.라벨링데이터\TL_질의응답데이터_안과"

json_files = [f for f in os.listdir(EVENT) if f.endswith(".json")]

data_list = []

for file in json_files:
    file_path = os.path.join(EVENT, file)

    with open(file_path, "r", encoding="utf-8") as f:
        data = json.load(f)

        # JSON 자체가 리스트라면
        if isinstance(data, list):
            data_list.extend(data)
        else:
            data_list.append(data)

df = pd.DataFrame(data_list)

print("JSON 파일 개수:", len(json_files))
print("DataFrame 행 개수:", len(df))
print(df.head())
print(df.columns)
```
