# training JSON 파일을 DataFrame으로 변환

`data/training` 아래의 모든 JSON 파일을 읽고, 중첩된 키를 컬럼으로 펼칩니다.

```python
from pathlib import Path
import json

import pandas as pd


DATA_DIR = Path("data/training")
json_files = sorted(DATA_DIR.rglob("*.json"))

records = []
failed_files = []

for file_path in json_files:
    try:
        with file_path.open("r", encoding="utf-8") as file:
            data = json.load(file)
    except (json.JSONDecodeError, OSError) as error:
        failed_files.append((file_path, error))
        continue

    if isinstance(data, list):
        records.extend(data)
    else:
        records.append(data)

if records:
    df = pd.json_normalize(records, sep=".")
else:
    df = pd.DataFrame()

print(f"JSON 파일 개수: {len(json_files):,}")
print(f"DataFrame 행 개수: {len(df):,}")
print(f"파싱 실패 파일 개수: {len(failed_files):,}")
print("컬럼 목록:", df.columns.tolist())

if failed_files:
    print("파싱 실패 파일:")
    for file_path, error in failed_files:
        print(f"- {file_path}: {error}")

df.head()
```

```python
df.to_csv('data/training/df.csv')
```
