```python
import pandas as pd
from pathlib import Path

csv_path = Path('../data/hospital.csv')
df = pd.read_csv(csv_path, encoding='utf-8')

address_column = "\ub3c4\ub85c\uba85\uc8fc\uc18c"
first_region_column = 'first_region'
second_region_column = 'second_region'
address_parts = df[address_column].astype('string').str.split()
df[first_region_column] = address_parts.str[0]
df[second_region_column] = address_parts.str[1]

# Map each first region to its unique second regions.
region_mapping = (
    df[[first_region_column, second_region_column]]
    .dropna()
    .drop_duplicates()
    .groupby(first_region_column)[second_region_column]
    .agg(list)
    .to_dict()
)

region_mapping
```

```python
import json
with open("../data/maping.txt", "w", encoding="utf-8") as f:
    json.dump(region_mapping, f, ensure_ascii=False, indent=4)
```

```python
assert isinstance(region_mapping, dict)
assert all(isinstance(second_regions, list) for second_regions in region_mapping.values())
assert region_mapping['\uc11c\uc6b8\ud2b9\ubcc4\uc2dc'][0] == '\uc885\ub85c\uad6c'
```
